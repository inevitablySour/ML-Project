```python
# =========================
# 0️⃣ Imports
# =========================
import pandas as pd
import numpy as np
import os
import re

# =========================
# 1️⃣ Load existing processed data
# =========================
data_path = "../phase2/processed_data/race_data_processed.csv"
df = pd.read_csv(data_path, parse_dates=['Date','birthdate'])
print(f"Original shape: {df.shape}")

# =========================
# 2️⃣ Fix age inconsistencies
# =========================
# Clip age_at_race to reasonable range
df['age_at_race'] = df['age_at_race'].clip(lower=15, upper=50)

# For rows with missing birthdate, fill age_at_race with Age
df.loc[df['birthdate'].isna(), 'age_at_race'] = df.loc[df['birthdate'].isna(), 'Age']

# =========================
# 3️⃣ Remove duplicate entries
# =========================
df = df.drop_duplicates(subset=['Rider','Race_ID','Stage_Number'])
print(f"Shape after removing duplicates: {df.shape}")

# =========================
# 4️⃣ Convert Timelag and Time to numeric seconds (safe version)
# =========================
def parse_timedelta_safe(t):
    if pd.isna(t):
        return 0
    t = str(t).strip().replace('+','').replace('-','').replace('–','')
    numbers = re.findall(r'\d+', t)
    if not numbers:
        return 0
    if len(numbers) == 3:
        return int(numbers[0])*3600 + int(numbers[1])*60 + int(numbers[2])
    elif len(numbers) == 2:
        return int(numbers[0])*60 + int(numbers[1])
    elif len(numbers) == 1:
        return int(numbers[0])
    return 0

df['Timelag_sec'] = df['Timelag'].apply(parse_timedelta_safe)
df['Time_sec'] = df['Time'].apply(parse_timedelta_safe)

print("Sample converted Timelag/Time:")
print(df[['Timelag','Timelag_sec','Time','Time_sec']].head(10))

# =========================
# 5️⃣ Optional: Fill missing categorical values
# =========================
df['Team'] = df['Team'].fillna('Unknown')
df['Stage_Name'] = df['Stage_Name'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')

# =========================
# 6️⃣ Save new processed CSV
# =========================
output_dir = "../phase2/processed_data"
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, "race_data_processed_v3.csv")

df.to_csv(output_path, index=False)
print(f"New processed data saved to: {output_path}")
print(f"New shape: {df.shape}")
```
